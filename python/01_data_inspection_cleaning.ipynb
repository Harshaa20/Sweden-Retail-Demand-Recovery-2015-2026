```python
import pandas as pd

file_1 = "/content/000006VS_20260904-213441.csv"
file_2 = "/content/000006VS_20260905-191336.csv"

df1 = pd.read_csv(file_1, skiprows=2)
df2 = pd.read_csv(file_2, skiprows=2)

print("FILE 1 shape:", df1.shape)
print("FILE 2 shape:", df2.shape)

print("\nFILE 1 — first 3 rows:")
display(df1.head(3))

print("\nFILE 2 — first 3 rows:")
display(df2.head(3))
```

```python
# Basic structural validation

print("FILE 1")
print("Categories:")
print(df1.iloc[:, 0].tolist())

print("\nNumber of columns:", len(df1.columns))

print("\nFILE 2")
print("Categories:")
print(df2.iloc[:, 0].tolist())

print("\nNumber of columns:", len(df2.columns))

# Check for missing/suppressed values
print("\nMissing/suppressed values:")
print("File 1:", (df1 == "..").sum().sum())
print("File 2:", (df2 == "..").sum().sum())

# Check duplicate column names
print("\nDuplicate column names:")
print("File 1:", df1.columns.duplicated().sum())
print("File 2:", df2.columns.duplicated().sum())
```

```python
# Compare overlapping columns between File 1 and File 2

overlap_columns = [col for col in df1.columns if col in df2.columns]

print("Number of overlapping columns:", len(overlap_columns))

# Compare the actual values
comparison = df1[overlap_columns].astype(str).equals(
    df2[overlap_columns].astype(str)
)

print("Are all overlapping values identical?:", comparison)
```

```python
# If they are not identical, identify where the differences are

if not comparison:
    differences = []

    for col in overlap_columns:
        if not df1[col].astype(str).equals(df2[col].astype(str)):
            differences.append(col)

    print("Number of columns with differences:", len(differences))
    print("\nFirst 20 columns with differences:")
    print(differences[:20])
```

```python
# Identify the three observation groups in the full historical file

observation_types = {}

for col in df2.columns[1:]:
    observation_type = col.rsplit(" ", 1)[0]
    observation_types[observation_type] = observation_types.get(observation_type, 0) + 1

for observation_type, count in observation_types.items():
    print(f"{observation_type}: {count} columns")
```

```python
# Show the first and last column belonging to each observation group

for observation_type in observation_types:
    matching_columns = [
        col for col in df2.columns
        if col.startswith(observation_type)
    ]

    print(f"\n{observation_type}")
    print("First:", matching_columns[0])
    print("Last: ", matching_columns[-1])
```

```python
# Reshape File 2 from wide format to long format

id_column = df2.columns[0]

df_long = df2.melt(
    id_vars=id_column,
    var_name="observation_month",
    value_name="value"
)

print("Long-format shape:", df_long.shape)

print("\nFirst 10 rows:")
display(df_long.head(10))
```

```python
# Split observation type and month

df_long["observation_type"] = df_long["observation_month"].str.extract(
    r"^(.*?)\s+(\d{4}M\d{2})$"
)[0]

df_long["month"] = df_long["observation_month"].str.extract(
    r"(\d{4}M\d{2})$"
)[0]

print("Observation types:")
print(df_long["observation_type"].unique())

print("\nNumber of observation types:", df_long["observation_type"].nunique())

print("\nNumber of months:", df_long["month"].nunique())

print("\nMonth range:")
print(df_long["month"].min(), "to", df_long["month"].max())

print("\nFirst 10 rows:")
display(df_long.head(10))
```

```python
print("Observation types:")
print(df_long["observation_type"].unique())

print("\nNumber of observation types:")
print(df_long["observation_type"].nunique())

print("\nNumber of months:")
print(df_long["month"].nunique())

print("\nMonth range:")
print(df_long["month"].min(), "to", df_long["month"].max())

print("\nMissing extracted values:")
print("Observation type:", df_long["observation_type"].isna().sum())
print("Month:", df_long["month"].isna().sum())
```

```python
# Locate all ".." values by category and observation type

missing_summary = (
    df_long[df_long["value"] == ".."]
    .groupby(["industrial classification (NACE Rev. 2)", "observation_type"])
    .size()
    .reset_index(name="missing_count")
)

display(missing_summary)
```

```python
# Check the earliest and latest month containing ".."

missing_rows = df_long[df_long["value"] == ".."]

print("Earliest month with '..':", missing_rows["month"].min())
print("Latest month with '..':", missing_rows["month"].max())
```

```python
# Inspect the missing months for the food category
# using the forecasting series only

forecast_series = "Seasonally adjusted, working day adjusted, constant prices"

food_missing = df_long[
    (df_long["industrial classification (NACE Rev. 2)"] ==
     "47.11+47.2 retail sale with mostly food") &
    (df_long["observation_type"] == forecast_series) &
    (df_long["value"] == "..")
]

print("Number of missing months:", len(food_missing))

print("\nFirst 20 missing months:")
print(food_missing["month"].head(20).tolist())

print("\nLast 20 missing months:")
print(food_missing["month"].tail(20).tolist())
```

```python
# Find the first and last available observation for every category and series

df_check = df_long.copy()

df_check["numeric_value"] = pd.to_numeric(
    df_check["value"],
    errors="coerce"
)

availability = (
    df_check[df_check["numeric_value"].notna()]
    .groupby([
        "industrial classification (NACE Rev. 2)",
        "observation_type"
    ])["month"]
    .agg(["min", "max", "count"])
    .reset_index()
)

display(availability)
```

```python
# Convert ".." to missing values and convert observations to numeric

df_long["value"] = pd.to_numeric(
    df_long["value"],
    errors="coerce"
)

print("Data type:", df_long["value"].dtype)

print("\nMissing values after conversion:", df_long["value"].isna().sum())

print("\nRemaining non-numeric values:")
print(df_long["value"].apply(lambda x: not pd.isna(x) and not isinstance(x, (int, float))).sum())
```

```python
# Create a proper monthly date and clean the category column name

df_long["date"] = pd.to_datetime(
    df_long["month"].str.replace("M", "-") + "-01"
)

df_long = df_long.rename(
    columns={
        "industrial classification (NACE Rev. 2)": "category"
    }
)

print("Date range:")
print(df_long["date"].min(), "to", df_long["date"].max())

print("\nDate data type:")
print(df_long["date"].dtype)

print("\nColumns:")
print(df_long.columns.tolist())

print("\nFirst 5 rows:")
display(df_long.head())
```

```python
# Final structural integrity check

expected_rows = 5 * 3 * 139

print("Actual rows:", len(df_long))
print("Expected rows:", expected_rows)

print("\nCategories:", df_long["category"].nunique())
print("Observation types:", df_long["observation_type"].nunique())
print("Months:", df_long["date"].nunique())

print("\nDuplicate category-series-month combinations:")
duplicates = df_long.duplicated(
    subset=["category", "observation_type", "date"]
).sum()
print(duplicates)

print("\nRows per observation type:")
print(df_long.groupby("observation_type").size())

print("\nRows per category:")
print(df_long.groupby("category").size())
```

```python
# Finalize the clean analytical dataset

df_clean = df_long.drop(
    columns=["observation_month"]
).copy()

# Create the three analytical datasets

df_forecast = df_clean[
    df_clean["observation_type"] ==
    "Seasonally adjusted, working day adjusted, constant prices"
].copy()

df_constant = df_clean[
    df_clean["observation_type"] ==
    "Constant prices"
].copy()

df_yoy = df_clean[
    df_clean["observation_type"] ==
    "Yearly development in percent, working day adjusted, constant prices"
].copy()

print("CLEAN DATASET")
print("Shape:", df_clean.shape)
print("Columns:", df_clean.columns.tolist())

print("\nFORECAST DATASET")
print("Shape:", df_forecast.shape)
print("Missing values:", df_forecast["value"].isna().sum())

print("\nCONSTANT-PRICE DATASET")
print("Shape:", df_constant.shape)
print("Missing values:", df_constant["value"].isna().sum())

print("\nYOY DATASET")
print("Shape:", df_yoy.shape)
print("Missing values:", df_yoy["value"].isna().sum())
```
